# LLE

머신러닝에서의 LLE(Locally Linear Embedding, 지역 선형 임베딩)는 고차원 데이터를 저차원으로 축소하는 대표적인 비선형 차원 축소(Non-linear Dimensionality Reduction) 알고리즘입니다.
기존의 PCA(주성분 분석)처럼 단순히 데이터를 평면에 투영하는 방식이 아니라, 데이터가 가진 고유한 기하학적 구조(매니폴드)를 보존하며 펼치는 매니폴드 학습(Manifold Learning)의 일종입니다.


### 💡 LLE의 핵심 아이디어

```티스토리
"고차원 공간에서 서로 가까이 있던 이웃들은 저차원 공간으로 내려와서도 여전히 가까운 이웃이어야 한다." 
```

LLE는 고차원에서는 구불구불하게 휜 복잡한 데이터라도, 아주 좁은 지역(Local)만 잘라놓고 보면 평면(선형, Linear)에 가깝다는 가정에서 출발합니다. 따라서 각 데이터 포인트를 주변 이웃들의 선형 결합(Linear combination)으로 나타내고, 이 관계를 저차원에서도 그대로 유지하도록 만듭니다. 

### 🛠️ LLE의 3단계 작동 과정
* 이웃 선택 (Select Neighbors)
  * 각 데이터 포인트마다 가장 가까운 *K*개의 이웃(K-Nearest Neighbors)을 찾습니다. (*K*값은 사용자가 지정하는 하이퍼파라미터입니다)
* 가중치 계산 (Reconstruct with Linear Weights)
  * 각 데이터를 이웃들의 결합으로 가장 잘 설명할 수 있는 가중치(*W*)를 계산합니다.
  * 이 가중치들은 원래 고차원 공간의 지역적 기하학 구조를 담고 있습니다.
* 저차원 매핑 (Map to Embedded Coordinates)
  * 고차원에서 구한 가중치(*W*)를 최대한 보존하면서 데이터를 저차원 좌표로 이동(임베딩)시킵니다. 

### ⚖️ LLE의 장점과 단점

## 👍 장점

* 비선형 데이터에 강력함: 스위스 롤(Swiss Roll) 데이터를 전래동요 멍석 말듯 말아놓은 형태처럼 선형 분리가 불가능한 데이터 분포를 평면으로 완벽하게 펼쳐낼 수 있습니다.
* 최적 해 보장: 복잡한 반복 계산이 아닌 선형 대수의 고유값 분해(Eigenvalue Decomposition) 문제를 풀기 때문에 지역 최솟값(Local Minima)에 빠지지 않고 언제나 최적의 해를 찾아냅니다. (Stanford Computer Graphics)

## 👎 단점
* 노이즈에 취약함: 데이터에 노이즈가 많거나 이상치(Outlier)가 있으면 이웃 관계가 왜곡되어 차원 축소 결과가 크게 망가집니다.
* 대용량 데이터에서 속도 저하: 모든 데이터 포인트마다 이웃을 찾고 거리를 계산해야 하므로 연산량이 큽니다.
* 하이퍼파라미터 민감도: 이웃의 개수(*K*)를 너무 작게 잡으면 데이터가 조각나고, 너무 크게 잡으면 비선형적 특성을 잃어버리게 됩니다.
