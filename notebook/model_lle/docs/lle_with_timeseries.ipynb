# LLE with Timeseries

LLE는 시계열 데이터(Time-Series Data)에도 충분히 사용이 가능합니다.

다만 시계열 데이터는 일반적인 정형 데이터와 달리 시간의 흐름에 따른 순서와 연속성(Temporal Dependency)이 중요하기 때문에, LLE를 그대로 적용하기보다는 데이터를 전처리하거나 변형하여 적용하는 것이 일반적입니다. 

시계열 데이터에서 LLE를 활용하는 대표적인 방법과 주의할 점을 정리해 드립니다. 

## 1. 시계열 데이터에서 LLE를 사용하는 3가지 방법

### 지연 임베딩(Lag/Delay Embedding) 결합
* 하나의 긴 시계열 데이터를 일정 시간 창(Window) 크기로 잘라 고차원 벡터로 만듭니다.
* 예를 들어, 최근 5일간의 주가 데이터를 하나의 5차원 데이터 포인트로 취급하는 방식입니다. 이렇게 생성된 고차원 데이터들을 LLE를 통해 2D나 3D로 축소하여 시스템의 동역학적 상태 변화를 시각화하거나 분석합니다.

### 시계열 특징(Feature) 기반 축소
* 시계열 데이터에서 평균, 분산, 왜도, 첨도, 트렌드, 계절성 등 다양한 통계적 특징들을 수십 개 추출합니다.
* 추출된 고차원 특징 벡터들에 LLE를 적용하여 데이터의 핵심 구조만 남기고 차원을 줄인 뒤, 군집화(Clustering)나 분류(Classification) 모델의 입력값으로 사용합니다.

### 다변량 시계열(Multivariate Time Series)의 차원 축소
* 수많은 센서에서 동시에 수집되는 데이터나 수백 개의 주식 종목처럼 변수가 너무 많은 다변량 시계열에서, 변수 간의 비선형적 관계를 유지하며 변수의 개수 자체를 줄일 때 LLE를 사용합니다. (ResearchGate)

## 2. ⚠️ 시계열 데이터 적용 시 주의할 점 및 한계
* 시간 순서의 무시: LLE는 기본적으로 데이터 간의 '공간적 거리(유클리드 거리 등)'를 기준으로 이웃을 찾습니다. 따라서 단순히 시점별 데이터를 LLE에 넣으면 시간의 전후 관계(순서) 정보가 완전히 유실됩니다.
* 거리 측정의 한계: 시계열 데이터는 길이가 다르거나 시간에 따라 밀리고 당겨지는 특성(위상차)이 있습니다. 일반적인 유클리드 거리를 사용하면 LLE가 이웃을 잘못 지정할 확률이 높으므로, DTW(Dynamic Time Warping) 같은 시계열 특화 거리 측정 방식을 결합하는 것이 좋습니다. (Dev Genius)

### 💡 요약 및 추천
* 시계열 데이터의 시각화나 복잡한 시스템의 상태 분석(매니폴드 분석) 목적이라면 LLE나 Isomap 같은 알고리즘이 훌륭한 도구가 될 수 있습니다.
* 만약 시계열의 순차적 정보 보존이 최우선이거나 대용량 시계열 예측/분류가 목적이라면, LLE보다는 시계열에 특화된 RNN, LSTM 혹은 시계열 오토인코더(Time-Series Autoencoder) 같은 딥러닝 기반의 임베딩 기법을 사용하는 것이 훨씬 더 효과적입니다
