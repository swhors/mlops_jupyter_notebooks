## 1. 열풍건조기에서 수집되는 센서 데이터

열풍 건조기(Hot Air Dryer)에서 수집하는 주요 센서 데이터 목록입니다. 건조기는 건조 대상물(농산물, 식품, 화학 제품 등)의 상태와 장비의 효율성을 실시간으로 관리하기 위해 다양한 센서를 활용합니다.

### **1. 핵심 환경 및 상태 데이터**

가장 기본적으로 수집되는 데이터는 건조 내부의 환경 조건입니다.

- **내부 온도:** 열풍의 설정 온도와 실제 내부 온도를 비교하여 히터를 제어합니다.
- **상대 습도:** 건조물의 수분이 빠져나오는 정도를 파악하여 배기 댐퍼의 개폐 여부를 결정합니다.
- **건조물 무게 (중량):** 건조 전후의 무게 변화를 통해 실시간 함수율(수분 함량)을 계산합니다.
- **표면 온도 (적외선):** 건조 대상물의 표면 온도를 직접 측정하여 과열에 의한 품질 저하를 방지합니다.

### **2. 기기 작동 및 제어 데이터**

장비의 안정적인 작동을 모니터링하기 위한 데이터입니다.

- **송풍기(Fan) 회전수(RPM):** 공기 순환 속도를 제어하여 건조 효율을 조절합니다.
- **전력 소비량:** 히터와 모터의 실시간 전력 사용량을 통해 에너지 효율과 과부하 상태를 체크합니다.
- **히터 작동 상태:** 히터의 On/Off 주기나 출력 비율(%) 데이터를 수집합니다.
- **배기압/풍압:** 필터 막힘이나 공기 흐름의 이상 유무를 확인하는 압력 센서 데이터입니다.

### **3. 안전 및 진단 데이터**

이상 징후를 감지하거나 장비를 보호하기 위해 수집합니다.

- **과승 방지 온도:** 시스템 오작동 시 화재 예방을 위해 독립적인 센서로 수집하는 최고 온도값입니다.
- **모터 진동/소음:** 팬 모터의 베어링 마모나 불균형을 사전에 감지하기 위한 가속도 데이터입니다.
- **도어 개폐 상태:** 작업 안전 및 열 손실 방지를 위해 수집되는 접점 데이터입니다.

### 요약 데이터 테이블

|구분|주요 수집 데이터 항목|활용 목적|
|------|---|---|
|품질 제어|내부 온·습도, 대상물 표면 온도, 함수율(중량)|정밀 건조 및 품질 균일화|
|운영 효율|소비 전력, 팬 RPM, 히터 가동률|에너지 절감 및 최적 운전 조건 도출|
|설비 보전|풍압, 모터 진동, 배기 온도|고장 예측(예후 진단) 및 부품 교체 주기 파악|
|안전 관리|과승 제한 온도, 도어 상태 센서|화재 예방 및 작업자 보호|
